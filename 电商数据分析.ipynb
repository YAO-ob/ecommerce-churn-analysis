# 数据清洗

```python
# 创建原始数据备份
df_original=df.copy()
print(f"备份数据形状：{df_original.shape}")
```

```python
# 处理缺失值

# 检查缺失情况
missing_summary={}
for col in df.columns:
    missing_count=df[col].isnull().sum()
    if missing_count>0:
        missing_summary[col]={
            "缺失数量":missing_count,
            "缺失比例%":(missing_count/len(df)*100).round(2),
            "数据类型":df[col].dtype
        }

# 处理缺失值
if missing_summary:
    missing_df=pd.DataFrame(missing_summary).T
    print("缺失值统计：")
    display(missing_df)

    # 根据数据类型以及缺失比例处理缺失值
    for col,info in missing_summary.items():
        dtype=info["数据类型"]
        missing_pct=info["缺失比例%"]

        # 缺失比例小于5%，删除缺失行
        if missing_pct<5:
            print(f"{col}:缺失比例{missing_pct}%，小于5%，删除缺失行")
            df=df.dropna(subset=[col])# 删除该列缺失值所在的行

        # 数值值：用于中位数进行填充
        elif dtype in ["int64","float64"]:
            median_val=df[col].median()# 计算中位数
            print(f"{col}:数值型变量，使用中位数{median_val}进行填充")
            df[col]=df[col].fillna(median_val)

        # 分类型：用众数进行填充
        elif dtype=="object":
            mode_val=df[col].mode()[0] if not df[col].mode().empty else "未知"
            print(f"{col}:分类型变量，使用众数{mode_val}进行填充")
            df[col]=df[col].fillna(mode_val)

else:
    print("数据当中没有缺失值！")

print(f"\n处理缺失值后数据形状为：{df.shape}")

    
```

```python
# 处理异常值

# 设置中文字体
plt.rcParams['font.sans-serif']=['SimHei']
plt.rcParams['axes.unicode_minus']=False

# 筛选数值类型列
numberic_cols=df.select_dtypes(include=[np.number]).columns.tolist()

if numberic_cols:
    outlier_info={}

    for col in numberic_cols:
        Q1=df[col].quantile(0.25)# 计算第一四分位数
        Q3=df[col].quantile(0.75)# 计算第三四分位数
        IQR=Q3-Q1
        # 获取上下限
        lower_bound=Q1-1.5*IQR
        upper_bound=Q3+1.5*IQR

        # 检查异常值（超出上下限的值）
        outliers=df[(df[col]<lower_bound)|(df[col]>upper_bound)]
        outlier_count=len(outliers)

        if outlier_count>0:
            outlier_info[col]={
                "异常值数量":outlier_count,
                "异常值比例%":round(outlier_count/len(df)*100,2),
                "最小值":df[col].min(),
                "第一四分位数Q1":Q1,
                "中位数":df[col].median(),
                "第三四分位数Q3":Q3,
                "最大值":df[col].max(),
                "下限":lower_bound,
                "上限":upper_bound
            }

    if outlier_info:
        outlier_df=pd.DataFrame(outlier_info).T
        print("异常值检测结果：")
        display(outlier_df)

        # 可视化处理
        fig,axes=plt.subplots(1,2,figsize=(16,6))

        # 左侧：异常值数量条形图
        outlier_count=[info["异常值数量"] for info in outlier_info.values()]
        cols_with_outliers=list(outlier_info.keys())

        axes[0].bar(cols_with_outliers,outlier_count,color="salmon")
        axes[0].set_title("各列异常值数量",fontsize=14,fontweight="bold")
        axes[0].set_xlabel("列名",fontsize=12)
        axes[0].set_ylabel("异常值数量",fontsize=12)
        axes[0].tick_params(axis="x",rotation=45)

        # 在条形图上添加具体数值标签
        for i,v in enumerate(outlier_count):
            axes[0].text(i,v+0.5,str(v),ha="center",va="bottom",fontsize=10)

        # 右侧使用箱线图（选择一个有异常值的列）
        example_col=cols_with_outliers[0] if cols_with_outliers else numberic_cols[0]
        axes[1].boxplot(df[example_col].dropna(),vert=False)
        axes[1].set_title(f"{example_col}的箱线图",fontsize=14,fontweight="bold")
        axes[1].set_xlabel(example_col,fontsize=12)

        plt.tight_layout()
        plt.show()

        # 使用缩尾法进行处理
        print("\n使用缩尾法进行处理")
        for col in cols_with_outliers:
            Q1=df[col].quantile(0.01)
            Q3=df[col].quantile(0.99)

            # 将超出1%与99%的值替换成边界值，进行缩尾
            df[col]=np.clip(df[col],Q1,Q3)
            print(f"{col}：将值限制在[{Q1:.2f}, {Q3:.2f}]范围内")
    else:
        print("数值型列当中没有检测到异常")

else:
    print("数据中没有数值型变量")
```

```python
# 创建衍生特征

# 活跃用户：距上次下单天数小于 30 天视为活跃（30 天为常用活跃口径）
df["是否活跃用户"]=df["距上次下单天数"].apply(lambda x:"活跃" if x<30 else "不活跃")

# 消费水平：按上月平均返现金额人工分档，<100 低消费，100-200 中消费，>=200 高消费
def get_consumption_level(amount):
    if amount<100:
        return "低消费"
    elif amount<200:
        return "中消费"
    else:
        return "高消费"

df["消费水平"]=df["上月平均返现金额"].apply(get_consumption_level)

# 使用频率：按上月订单数量分档，0 无订单，1-2 低频，3-5 中频，>5 高频
def get_frequency_level(count):
    if count==0:
        return "无订单"
    elif count<=2:
        return "低频"
    elif count<=5:
        return "中频"
    else:
        return "高频"

df["使用频率"]=df["上月订单数量"].apply(get_frequency_level)

# 满意度分组：评分 1-2 不满意，3-4 一般，5 满意
def get_satisfaction_group(score):
    if score<=2:
        return "不满意"
    elif score<=4:
            return "一般"
    else:
        return "满意"

df["满意度分组"]=df["服务满意度评分"].apply(get_satisfaction_group)

print(f"数据形状{df.shape}")
display(df.head())
```

## 流失用户特征分析

```python
# 构造二值流失标签
df['是否流失'] = (df['流失标志'] == '流失').astype(int)
overall_churn = df['是否流失'].mean()
print(f'整体流失率：{overall_churn*100:.2f}%（样本 {len(df)} 条）')

def churn_by(col):
    """按指定分组统计人数、流失数与流失率，按流失率降序"""
    g = df.groupby(col).agg(人数=('是否流失', 'count'), 流失数=('是否流失', 'sum'))
    g['流失率%'] = (g['流失数'] / g['人数'] * 100).round(1)
    return g.sort_values('流失率%', ascending=False)

# 逐个分类维度输出流失率对比
cat_cols = ['性别', '城市等级', '婚姻状况', '年龄分组', '首选登录设备',
            '是否活跃用户', '消费水平', '使用频率', '满意度分组']
for col in cat_cols:
    if col in df.columns:
        print('\n====', col, '====')
        print(churn_by(col).to_string())
```

```python
# 以婚姻状况为例绘制流失率对比条形图（可将 churn_by 的参数换成其他维度）
view = churn_by('婚姻状况')
fig, ax = plt.subplots(figsize=(8, 4))
ax.bar(view.index, view['流失率%'], color='steelblue')
ax.axhline(overall_churn * 100, color='red', linestyle='--',
           label=f'整体流失率 {overall_churn*100:.1f}%')
ax.set_title('不同婚姻状况客户流失率对比')
ax.set_ylabel('流失率 %')
for i, v in enumerate(view['流失率%']):
    ax.text(i, v + 0.5, f'{v}%', ha='center')
ax.legend()
plt.tight_layout()
plt.show()
```

```python
# 数值维度：流失组与未流失组中位数对比
num_cols = ['使用平台时长（月）', 'APP使用时长（小时）', '上月订单数量',
            '距上次下单天数', '服务满意度评分']
rows = []
for c in num_cols:
    if c in df.columns:
        rows.append({
            '字段': c,
            '流失组中位数': df[df['是否流失'] == 1][c].median(),
            '未流失组中位数': df[df['是否流失'] == 0][c].median()
        })
print(pd.DataFrame(rows).to_string(index=False))
```
